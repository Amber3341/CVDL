```python
import torch
import cv2
import time
import numpy as np
from pathlib import Path

# Load the YOLOv5 model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

# Path to input video
video_path = '/kaggle/input/videod/2103099-uhd_3840_2160_30fps.mp4'  # Change this to your video file
if not Path(video_path).exists():
    raise FileNotFoundError("Error: input.mp4 not found. Please upload a valid video file.")

cap = cv2.VideoCapture(video_path)

# Get video properties
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Define codec and create VideoWriter object
out = cv2.VideoWriter('output.mp4', cv2.VideoWriter_fourcc(*'mp4v'), 20.0, (frame_width, frame_height))

if not out.isOpened():
    raise Exception("Error: VideoWriter failed to initialize.")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    
    start_time = time.time()  # Start time for FPS calculation
    
    # Convert frame to YOLO format
    results = model(frame)
    
    # Get processed frame with detections
    output_frame = np.squeeze(results.render())
    
    # Calculate FPS
    end_time = time.time()
    fps_text = f'FPS: {1 / (end_time - start_time):.2f}'
    
    # Display FPS on the frame
    cv2.putText(output_frame, fps_text, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    
    # Write frame to output video
    out.write(output_frame)
    print("Processing frame...")
    
cap.release()
out.release()
cv2.destroyAllWindows()

# Verify if output.mp4 was saved
if Path('output.mp4').exists():
    print("Processing complete. Output saved as output.mp4")
else:
    print("Error: output.mp4 was not saved correctly.")
```
